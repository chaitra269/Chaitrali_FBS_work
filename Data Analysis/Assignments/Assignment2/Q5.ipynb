Q.5. Create a 3x3 array of integers and access:
1. The second element of the first row.
2. The third column of the 2nd row.
3. A subarray with the first two rows and last two columns.

```python
import numpy as np
```

```python
arr = np.arange(9).reshape(3,3)
arr
```

```python
a1 = arr[0,1]
a1
```

```python
a2 = arr[1,2]
a2
```

```python
a3 = arr[0:2,1:3]
a3
```
