Q.9. Use NumPy's built-in functions to compute the sum, mean, and standard deviation of a
given array.

```python
import numpy as np
```

```python
data = np.array([10, 20, 30, 40, 50])
data
```

```python
data_sum = np.sum(data)
data_sum
```

```python
data_mean = np.mean(data)
data_mean
```

```python
data_std = np.std(data)
data_std
```
