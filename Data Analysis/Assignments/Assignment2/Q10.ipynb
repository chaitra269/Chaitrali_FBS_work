Q.10. Create a 2x2 matrix A and a 2x2 matrix B. Perform matrix multiplication using np.dot() or
the @ operator.

```python
import numpy as np
```

```python
A = np.array([[1, 2], [3, 4]])
A
```

```python
B = np.array([[5, 6], [7, 8]])
B
```

```python
result = A @ B
result
```

```python
result = np.dot(A,B)
result
```
