Q.13. Create a 4x4 array of random integers. Find the maximum, minimum, and mean of the
entire array.

```python
import numpy as np
```

```python
arr = np.random.randint(1,23,size=(4,4))
arr
```

```python
arr_max = np.max(arr)
arr_max
```

```python
arr_min = np.min(arr)
arr_min
```

```python
arr_mean = np.mean(arr)
arr_mean
```
