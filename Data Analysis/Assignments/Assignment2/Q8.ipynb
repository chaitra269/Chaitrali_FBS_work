Q.8. Create two arrays of shape (2, 3): arr1 and arr2. Perform element-wise addition,
subtraction, multiplication, and division on them.

```python
import numpy as np
```

```python
arr1 = np.array([[10, 20, 30], [40, 50, 60]])
arr1
```

```python
arr2 = np.array([[2, 4, 5], [10, 2, 3]])
arr2
```

```python
arr3 = arr1 + arr2
arr3
```

```python
arr3 = arr1 - arr2
arr3
```

```python
arr3 = arr1 * arr2
arr3
```

```python
arr3 = arr1 / arr2
arr3
```
