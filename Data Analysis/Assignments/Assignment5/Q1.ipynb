Refer to the tips.csv dataset. The dataset contains total 244 tips(data about customers
served by a restaurant) across 7 different columns.
1. totbill - Total bill (cost of the meal), including tax, in US dollars
2. tip - Tip (gratuity) in US dollars
3. sex - Sex of person paying for the meal (0=male, 1=female)
4. smoker - Smoker in party? (0=No, 1=Yes)
5. day - Sun, Sat, Thu, Fri
6. time - Night and Day
7. size - Size of the party

Q.1. Plot a day wise boxplot of totbill column.

```python
import numpy as np
```

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/chait/OneDrive/Desktop/FBS/Data Analysis/Datasets/Datasets/tips.csv")
df
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='day', y='total_bill')
plt.title('Day-wise Boxplot of Total Bill')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

Q.2. Plot a pariplot by setting hue parameter t

```python
sns.pairplot(data=df, hue='time')
plt.suptitle('Pairplot of Dataset Variables', y=1.02)
plt.show()
```

Q.3. Draw a scatter plot of tip and totalbill and also allot the points according to the sex

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')
plt.title('Scatter Plot of Tip vs Total Bill by Sex')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

Q.4. Plot a hitsogram of totbill for 50 bins

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='total_bill', bins=50, kde=True)
plt.title('Histogram of Total Bill (50 Bins)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Count')
plt.show()
```

Q.5. Plot a heatmap to represent entire dataset's correlation

```python
plt.figure(figsize=(6, 4))
numerical_df = df[['total_bill', 'tip', 'sex', 'smoker', 'size']].copy()
numerical_df['sex']=numerical_df['sex'].map({'Male':0,'Female':1})
numerical_df['smoker']=numerical_df['smoker'].map({'No':0,'Yes':1})
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt=".2f",linewidths=0.5)
plt.title('Dataset Correlation Heatmap')
plt.show()
```
