#### Bar Chart:
1. Create a bar chart to compare the average price of cars by manufacturer.
2. Label the axes and add a suitable title.
3. Use different colors for different manufacturers.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
cars = pd.read_csv("C:/Users/chait/OneDrive/Desktop/FBS/Data Analysis/Datasets/Datasets/cars93.csv")
cars
```

```python
avg_price = cars.groupby('Manufacturer')['Price'].mean().reset_index()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=avg_price, x='Manufacturer', y='Price', hue='Manufacturer', palette='viridis', legend=False)
plt.xticks(rotation=90)
plt.xlabel("Manufacturer")
plt.ylabel("Average Price (in $1,000)")
plt.title("Average Car Price by Manufacturer")
plt.tight_layout()
plt.show()
```
