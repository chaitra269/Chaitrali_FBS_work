#### Scatter Plot:
1. Create a scatter plot to show the relationship between horsepower and MPG
(Miles Per Gallon).
2. Label the axes and add a title.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
cars = pd.read_csv("C:/Users/chait/OneDrive/Desktop/FBS/Data Analysis/Datasets/Datasets/cars93.csv")
cars
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=cars, x='Horsepower', y='MPG.city', color='red')
plt.xlabel("Horsepower")
plt.ylabel("City MPG")
plt.title("Relationship Between Horsepower and City MPG")
plt.grid(True)
plt.show()
```
