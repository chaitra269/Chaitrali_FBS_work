#### Pie Chart:
1. Create a pie chart showing the proportion of different car types in the dataset.
2. Include percentages in the chart.

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
cars = pd.read_csv("C:/Users/chait/OneDrive/Desktop/FBS/Data Analysis/Datasets/Datasets/cars93.csv")
cars
```

```python
type_counts = cars['Type'].value_counts()
```

```python
plt.figure(figsize=(7, 7))
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Proportion of Different Car Types")
plt.show()
```
