Q.3. Draw a scatter plot of tip and totalbill and also allot the points according to the sex

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')
plt.title('Scatter Plot of Tip vs Total Bill by Sex')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

Q.4. Plot a hitsogram of totbill for 50 bins

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='total_bill', bins=50, kde=True)
plt.title('Histogram of Total Bill (50 Bins)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Count')
plt.show()
```

Q.5. Plot a heatmap to represent entire dataset's correlation

```python
plt.figure(figsize=(6, 4))
numerical_df = df[['total_bill', 'tip', 'sex', 'smoker', 'size']].copy()
numerical_df['sex']=numerical_df['sex'].map({'Male':0,'Female':1})
numerical_df['smoker']=numerical_df['smoker'].map({'No':0,'Yes':1})
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt=".2f",linewidths=0.5)
plt.title('Dataset Correlation Heatmap')
plt.show()
```
