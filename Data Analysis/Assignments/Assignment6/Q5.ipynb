#### Line Chart:
1. Create a line chart showing the trend of horsepower for different types of cars.
2. Include proper legends, labels, and a title

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
cars = pd.read_csv("C:/Users/chait/OneDrive/Desktop/FBS/Data Analysis/Datasets/Datasets/cars93.csv")
cars
```

```python
hp_trend = cars.groupby('Type')['Horsepower'].mean().reset_index()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(hp_trend['Type'], hp_trend['Horsepower'], marker='o', color='b', label='Avg Horsepower')
plt.xlabel("Car Type")
plt.ylabel("Horsepower")
plt.title("Trend of Horsepower by Car Type")
plt.legend()
plt.grid(True)
plt.show()
```
